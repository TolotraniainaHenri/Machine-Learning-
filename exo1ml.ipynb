```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

```python
# Creation de donnee 
x,y = make_regression(n_samples= 100, n_features= 1, noise= 12)
```

```python
# Transformation de y en matrice colonne
y = y.reshape(-1, 1)
```

```python
# Creation de la matrice X , Biais 
ones=np.ones((x.shape[0], 1))
```

```python
ones.shape
```

```python
X=np.hstack((x,ones))
```

```python
X.shape
```

```python
# Creation de model lineaire
theta = np.random.randn(2, 1)
def modele(X,theta):
    return X @ theta
```

```python
modele(X, theta).shape
```

```python
# Fonction cout
def foction_cout( X, theta, y):
    predictions = modele(X, theta)
    m = len(y)
    return (1/(2*m))*np.sum(( predictions - y)**2)
```

```python
# Gradient
def gradient(X, theta, y):
    m = len(y)

    return (1 / m) * X.T @ (modele(X, theta) - y)
```

```python
gradient( X, theta, y).shape
```

```python
# Descente de gradient
def descente_gradient(X, y, theta, alpha, n_iteration):
    for i in range(n_iteration):
        theta = theta - alpha * gradient( X, theta, y)
    return theta 
```

```python
# Resutal
new_theta = descente_gradient(X, y, theta, 0.01, 1000)
```

```python
new_theta.shape
```

```python
theta.shape
```

```python
# Aprentissage  Avant 
print('Ancien theta ',theta)
#print('Nouveau theta ', new_theta)
print('Coup initiale :', foction_cout( X, theta, y))
print('Coup final:', foction_cout( X, new_theta, y))
```

```python
plt.scatter(x, y , label = "Donnee reel")
plt.plot(x, modele(X, theta), color = "red")
```

```python
plt.scatter(x, y , label = "Donnee reel")
plt.plot(x, modele(X,theta), color = "red" , label = "Droite de regretion initial")
plt.plot(x, modele(X, new_theta), color = "blue", label = "Droite de regression")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
```
